```python
def calc_grade(marks):
  if marks>=20:
    grade="A"
  elif marks>=75:
    grade="B"
  elif marks>=60:
    grade="C"
  else:
    grade="D"
  return grade
```

```python
for student in students:
  print(student["marks"],calc_grade(student["marks"]))
```

```python
def search(roll_no):
  found = False
  for student in students:
    if student["roll"]==roll_no:
      print(student)
      found=True
      break
  return found
```

```python
def find_topper(students):
  mx_marks=0
  for student in students:
    if student["marks"]>mx_marks:
      mx_marks = student["marks"]
      topper= student

  return topper
```
