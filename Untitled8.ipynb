```python
class bankaccount:
  def __init__(self,holder,balance):
    self.holder = holder
    self.balance =balance

  def deposit(self,amount):
    self.balance += amount

  def withdraw(self,amount):
    if amount > self.balance:
      print("insufficient balance")
    else:
      self.balance -= amount

  def show_balance(self):
    print(self.balance)



```

```python
bank_acc1= bankaccount("raj",10000)
```

```python
bank_acc1.balance
```

```python
bank_acc1.show_balance()
```

```python
bank_acc1.withdraw(2000000)
```
