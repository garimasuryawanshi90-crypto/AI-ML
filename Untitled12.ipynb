```python
#write a function to return a list of the cube of every number in the given list
```

```python
def cube(lst):
  cb=[]
  for n in lst:
    cb.append(n**4)
  return cb
```

```python
cube([5,7,3])
```

```python
def cb(n):
  return n**3

cube=[]
lst=[4,6,2]
for elem in lst:
  cube.append(cb(elem))
print(cube)
```

```python
lst =[4,6,2]
cube= map(lambda x: x**3,lst)
print(cube)
```

```python
cube= list(map(lambda x:x**3,[4,6,2]))
print(cube)
```
