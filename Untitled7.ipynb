```python
s=input()
sl=s.lower()
vowel=0

consonent=0
digits=0
spaces=0

for ch in sl:
  if ch in "aeiou":
   vowel+=1
  elif ch >="a"and ch<="z":
    consonent+=1
  elif ch>="0" and ch<="z":
     digits+=1
  elif ch==" ":
    spaces+=1
  print("vowel-", vowel)
  print("consonent-", consonent)
  print("digits-" ,digits)
  print("spaces-", spaces)
```
